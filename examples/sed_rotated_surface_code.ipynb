```python
%matplotlib inline
import matplotlib.pyplot as plt

from dizx.code import rotated_surface_code
from dizx.zx_diagram import SEDiagram
```

```python
code = rotated_surface_code(3)
se = SEDiagram.from_code(code, rounds=3)
code, len(se.detectors), len(se.stabilisers), len(se.co_stabilisers), len(se.logicals), se.webs_complete()
```

```python
se.draw(title="SED: d=3 rotated surface code, 3 rounds")
plt.show()
```

```python
for k, w in enumerate(se.detectors):
    se.draw(w, title=f"detector {k}")
    plt.show()
```

```python
for k, w in enumerate(se.stabilisers):
    se.draw(w, title=f"stabiliser {k}: {w.boundary()[1]}")
    plt.show()
```

```python
for k, w in enumerate(se.co_stabilisers):
    se.draw(w, title=f"co-stabiliser {k}: {w.boundary()[0]}")
    plt.show()
```

```python
for k, w in enumerate(se.logicals):
    se.draw(w, title=f"logical {k}: {w.boundary()[0]} -> {w.boundary()[1]}")
    plt.show()
```

```python
from dizx.realisation import RealisedSED

real = RealisedSED.circuit_based(se)
real.draw(title="circuit-based realisation")
plt.show()
```

```python
for name, w in [("detector 4", real.detectors[4]), ("stabiliser 0", real.stabilisers[0]),
                ("co-stabiliser 0", real.co_stabilisers[0]), ("logical 0", real.logicals[0])]:
    real.draw(w, title=f"{name}: outcomes {sorted(real.record(w))}")
    plt.show()
```
